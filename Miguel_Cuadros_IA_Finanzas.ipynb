## Paso 1. Entender el problema

El proyecto busca apoyar la gestion de finanzas personales mediante un modelo que identifique patrones de gasto que pueden necesitar mayor revision.

Para este Sprint 1 trabajaremos con una clasificacion multiclase:

- `gasto_controlado`
- `gasto_frecuente`
- `gasto_elevado`
- `gasto_no_esencial_alto`
- `gasto_inusual`
- `gasto_emergencia`
- `gasto_incremento_recurrente`

Esta primera version usa reglas explicables para crear etiquetas iniciales y luego entrena un modelo sobre esas etiquetas.


## Paso 2. Importar librerias

Usamos `pandas` para datos y `scikit-learn` para el modelo de Machine Learning.


```python
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import MinMaxScaler, StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

pd.set_option("display.max_columns", None)
```

## Paso 3. Cargar dataset ampliado

El dataset base fue ampliado a 500 registros para que el modelo tenga mas ejemplos de gastos comunes.


```python
ruta = "data/raw/gastos_semana_1.csv"
df = pd.read_csv(ruta)

print(f"Filas: {df.shape[0]} | Columnas: {df.shape[1]}")
df.head()
```

## Paso 4. Revisar estructura del dataset

Verificamos columnas, valores nulos y distribucion por categoria.


```python
print("Columnas:")
print(df.columns.tolist())

print("\nTipos de datos:")
print(df.dtypes)

print("\nValores nulos:")
print(df.isnull().sum())

print("\nCantidad de gastos por categoria:")
print(df["categoria"].value_counts())
```

## 5. Ingenieria de caracteristicas

En esta parte transformamos los datos crudos en variables que el modelo pueda entender.

Creamos:

- `mes`
- `peso_categoria`
- `frecuencia`
- `porcentaje_categoria`
- `alerta_incremento`
- `monto_normalizado`
- `frecuencia_normalizada`
- `score_financiero`


```python
mapa_pesos = {
    "servicios": 1.0,
    "transporte": 1.1,
    "educacion": 1.1,
    "alimentacion": 1.2,
    "salud": 1.3,
    "ocio": 1.8,
    "entretenimiento": 1.9,
    "compras_no_esenciales": 2.0,
}

df = df.copy()
df["fecha"] = pd.to_datetime(df["fecha"], format="%Y-%m-%d")
df = df.sort_values("fecha").reset_index(drop=True)
df["mes"] = df["fecha"].dt.month
df["peso_categoria"] = df["categoria"].map(mapa_pesos)
df["frecuencia"] = df.groupby("categoria")["categoria"].transform("count")
df["porcentaje_categoria"] = df["frecuencia"] / len(df)
df["monto_previo"] = df.groupby("descripcion")["monto"].shift(1)
df["alerta_incremento"] = (
    df["monto_previo"].notna() & (df["monto"] > df["monto_previo"] * 1.10)
).astype(int)

scaler_score = MinMaxScaler()
df[["monto_normalizado", "frecuencia_normalizada"]] = scaler_score.fit_transform(
    df[["monto", "frecuencia"]]
)

df["score_financiero"] = (
    df["monto_normalizado"]
    + df["peso_categoria"]
    + df["frecuencia_normalizada"]
    + df["alerta_incremento"]
)

columnas_trabajo = [
    "fecha",
    "descripcion",
    "categoria",
    "monto",
    "mes",
    "peso_categoria",
    "frecuencia",
    "porcentaje_categoria",
    "alerta_incremento",
    "monto_normalizado",
    "frecuencia_normalizada",
    "score_financiero",
]

df[columnas_trabajo].head(10)
```

## Paso 6. Crear etiquetas multiclase

Construimos una variable llamada `tipo_gasto`.

Reglas usadas:

- `gasto_emergencia`: gasto alto de salud.
- `gasto_incremento_recurrente`: gasto recurrente que subio mas de 10 por ciento frente al registro anterior.
- `gasto_inusual`: gasto de monto relevante en una categoria que representa menos del 15 por ciento del dataset.
- `gasto_no_esencial_alto`: gasto alto en ocio, entretenimiento o compras no esenciales.
- `gasto_elevado`: gasto alto que no cae en las reglas anteriores.
- `gasto_frecuente`: gasto de monto bajo o medio en una categoria muy repetida.
- `gasto_controlado`: gasto sin senales fuertes de alerta.

La etiqueta `gasto_incremento_recurrente` se agrega porque ayuda a detectar gastos fijos o repetidos que empiezan a subir, incluso si su categoria no es riesgosa por si sola.


```python
umbral_monto_alto = df["monto"].quantile(0.75)
umbral_monto_medio = df["monto"].quantile(0.60)
umbral_monto_bajo = df["monto"].quantile(0.35)
umbral_frecuencia_alta = df["frecuencia"].quantile(0.75)
categorias_no_esenciales = ["ocio", "entretenimiento", "compras_no_esenciales"]

def asignar_tipo_gasto(row):
    if row["categoria"] == "salud" and row["monto"] >= umbral_monto_alto:
        return "gasto_emergencia"
    if row["alerta_incremento"] == 1:
        return "gasto_incremento_recurrente"
    if row["porcentaje_categoria"] < 0.15 and row["monto"] >= umbral_monto_alto:
        return "gasto_inusual"
    if row["categoria"] in categorias_no_esenciales and row["monto"] >= umbral_monto_medio:
        return "gasto_no_esencial_alto"
    if row["monto"] >= umbral_monto_alto:
        return "gasto_elevado"
    if row["frecuencia"] >= umbral_frecuencia_alta and row["monto"] <= umbral_monto_bajo:
        return "gasto_frecuente"
    return "gasto_controlado"

df["tipo_gasto"] = df.apply(asignar_tipo_gasto, axis=1)
df["requiere_atencion"] = (df["tipo_gasto"] != "gasto_controlado").astype(int)

print("Umbral monto alto:", round(umbral_monto_alto, 2))
print("Umbral monto medio:", round(umbral_monto_medio, 2))
print("Umbral monto bajo:", round(umbral_monto_bajo, 2))
print("Umbral frecuencia alta:", round(umbral_frecuencia_alta, 2))

print("\nDistribucion de tipo_gasto:")
print(df["tipo_gasto"].value_counts())

df[["descripcion", "categoria", "monto", "porcentaje_categoria", "score_financiero", "tipo_gasto"]].head(15)

```

## Paso 7. Analisis breve antes del modelo

Revisamos montos por categoria y los gastos con mayor score financiero.


```python
resumen_categoria = (
    df.groupby("categoria")
    .agg(
        cantidad=("categoria", "count"),
        porcentaje=("porcentaje_categoria", "first"),
        monto_total=("monto", "sum"),
        score_promedio=("score_financiero", "mean"),
    )
    .sort_values("monto_total", ascending=False)
)

display(resumen_categoria)

print("Top 10 gastos con mayor score financiero:")
display(
    df[["fecha", "descripcion", "categoria", "monto", "score_financiero", "tipo_gasto"]]
    .sort_values("score_financiero", ascending=False)
    .head(10)
)
```

## Paso 8. Definir variables para el modelo

Variables de entrada:

- `monto`
- `peso_categoria`
- `frecuencia`
- `porcentaje_categoria`
- `alerta_incremento`
- `monto_normalizado`
- `frecuencia_normalizada`
- `score_financiero`

Variable de salida:

- `tipo_gasto`


```python
features = [
    "monto",
    "peso_categoria",
    "frecuencia",
    "porcentaje_categoria",
    "alerta_incremento",
    "monto_normalizado",
    "frecuencia_normalizada",
    "score_financiero",
]

X = df[features]
y = df["tipo_gasto"]

print("Variables de entrada:")
display(X.head())

print("\nVariable objetivo:")
print(y.value_counts())
```

## Paso 9. Dividir datos en entrenamiento y prueba

Usamos 70 por ciento de los datos para entrenar y 30 por ciento para probar.


```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.30,
    random_state=42,
    stratify=y,
)

print(f"Datos de entrenamiento: {X_train.shape[0]}")
print(f"Datos de prueba: {X_test.shape[0]}")
```

## Paso 10. Entrenar el modelo

Usamos un `Pipeline` para escalar las variables y luego entrenar la regresion logistica multiclase.


```python
modelo = Pipeline(
    steps=[
        ("scaler", StandardScaler()),
        (
            "clasificador",
            LogisticRegression(
                random_state=42,
                max_iter=2000,
                class_weight="balanced",
            ),
        ),
    ]
)

modelo.fit(X_train, y_train)
print("Modelo multiclase entrenado correctamente.")
```

## Paso 11. Evaluar el modelo

Revisamos la precision, la matriz de confusion y el reporte de clasificacion.


```python
predicciones = modelo.predict(X_test)
precision = accuracy_score(y_test, predicciones)
clases = modelo.classes_
aciertos = (predicciones == y_test).sum()
total_pruebas = len(y_test)

print(f"Precision del modelo: {precision:.2f}")
print(f"Precision en porcentaje: {precision * 100:.2f}%")
print(f"Aciertos: {aciertos} de {total_pruebas} registros de prueba")

matriz = pd.DataFrame(
    confusion_matrix(y_test, predicciones, labels=clases),
    index=[f"Real {clase}" for clase in clases],
    columns=[f"Pred {clase}" for clase in clases],
)

display(matriz)

print("Reporte de clasificacion:")
print(
    classification_report(
        y_test,
        predicciones,
        labels=clases,
        zero_division=0,
    )
)
```

## Paso 12. Interpretar probabilidades

Ademas de predecir una clase, el modelo puede entregar la probabilidad mas alta asociada a su decision.


```python
resultados = X_test.copy()
resultados["real"] = y_test.values
resultados["prediccion"] = predicciones
resultados["probabilidad_maxima"] = modelo.predict_proba(X_test).max(axis=1)

resultados.sort_values("probabilidad_maxima", ascending=False).head(10)
```

## Paso 13. Probar con gastos nuevos

Creamos ejemplos nuevos para observar como responde el modelo ante cada tipo de gasto.


```python
nuevos_gastos = pd.DataFrame(
    [
        {"descripcion": "Pago estable de internet", "categoria": "servicios", "monto": 90000, "frecuencia": 20, "alerta_incremento": 0},
        {"descripcion": "Consulta medica urgente", "categoria": "salud", "monto": 220000, "frecuencia": 38, "alerta_incremento": 0},
        {"descripcion": "Servicio con incremento fuerte", "categoria": "servicios", "monto": 230000, "frecuencia": 125, "alerta_incremento": 1},
        {"descripcion": "Compra inusual de material", "categoria": "educacion", "monto": 240000, "frecuencia": 35, "alerta_incremento": 0},
        {"descripcion": "Compra no esencial costosa", "categoria": "compras_no_esenciales", "monto": 210000, "frecuencia": 16, "alerta_incremento": 0},
        {"descripcion": "Pago alto de mercado", "categoria": "alimentacion", "monto": 180000, "frecuencia": 154, "alerta_incremento": 0},
        {"descripcion": "Cafe diario fuera de casa", "categoria": "alimentacion", "monto": 9000, "frecuencia": 154, "alerta_incremento": 0},
    ]
)

nuevos_gastos["peso_categoria"] = nuevos_gastos["categoria"].map(mapa_pesos)
nuevos_gastos["porcentaje_categoria"] = nuevos_gastos["frecuencia"] / len(df)
nuevos_gastos[["monto_normalizado", "frecuencia_normalizada"]] = scaler_score.transform(
    nuevos_gastos[["monto", "frecuencia"]]
)
nuevos_gastos["score_financiero"] = (
    nuevos_gastos["monto_normalizado"]
    + nuevos_gastos["peso_categoria"]
    + nuevos_gastos["frecuencia_normalizada"]
    + nuevos_gastos["alerta_incremento"]
)

prediccion_nueva = modelo.predict(nuevos_gastos[features])
probabilidad_nueva = modelo.predict_proba(nuevos_gastos[features]).max(axis=1)

nuevos_gastos["prediccion_tipo_gasto"] = prediccion_nueva
nuevos_gastos["probabilidad_maxima"] = probabilidad_nueva

nuevos_gastos[[
    "descripcion",
    "categoria",
    "monto",
    "peso_categoria",
    "frecuencia",
    "porcentaje_categoria",
    "score_financiero",
    "prediccion_tipo_gasto",
    "probabilidad_maxima",
]]
```

```python
columnas_salida = [
    "fecha",
    "descripcion",
    "categoria",
    "monto",
    "mes",
    "peso_categoria",
    "frecuencia",
    "porcentaje_categoria",
    "alerta_incremento",
    "monto_normalizado",
    "frecuencia_normalizada",
    "score_financiero",
    "tipo_gasto",
    "requiere_atencion",
]

ruta_salida = "data/processed/gastos_sprint1_final.csv"
df[columnas_salida].to_csv(ruta_salida, index=False)
print(f"Dataset final guardado en: {ruta_salida}")
```
